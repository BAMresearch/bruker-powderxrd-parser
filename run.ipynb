```python
from bam_masterdata.logger import logger
from bam_masterdata.metadata.entities import CollectionType
from bam_masterdata.cli.run_parser import RunParsers
from pathlib import Path
from pybis import Openbis
from decouple import config as environ


coll = CollectionType()
```

Run File for a Bam-masterdata Parser

Step 1:
Import the Parser Class from the src folder.

```python
from src.bruker_powderxrd_parser import BrukerPowderXRDParser

parser = BrukerPowderXRDParser()
```

Declare the Files to parse

```python

FILE_1 = "tests\\data\\D6Pm\\A52-B128-S_20-140_0.008_4s_17h_D6Pm.brml"
#FILE_2 = ...
```

Step 3:
Create own .env file from .env.example
 - Copy and rename .env.example to .env
 - Set own username and password
 - Set Openbis URL

```python
URL = environ("OPENBIS_URL")
print(f"Connecting to openBIS instance at {URL}...")
o = Openbis(URL)
o.login(
    environ("OPENBIS_USERNAME"),
    environ("OPENBIS_PASSWORD"),
    save_token=True,
)
```

Step 4:
Define Space, Projekt and optionally Collection for OpenBis

```python
SPACE = "EXAMPLE_SPACE"
PROJECT = "EXAMPLE_PROJECT"
O_COLLECTION = "EXAMPLE_COLLECTION" # optional

#if not provided, the file are save in the project
```

Step 5:
Define Parser run.

```python
parsing = RunParsers(o,SPACE,PROJECT,O_COLLECTION,{parser: [FILE_1]},logger=logger,)
```

Step 6:
Starting the run.

```python

print("Starting parsing...")
parsing.run()
print("done")
```
